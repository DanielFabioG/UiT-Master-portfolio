### Formueeffekt som moderator

Dersom råvarepriser og KPI endrer seg avhengig av hvor sterkt boligmarkedet/gullprisen/wealth effekten er i en gitt periode, så kan dette være en moderator for effekten av råvarepriser og valutakurser på KPI-vekst. Siden den endrer styrken og retningen av denne effekten.


### Formueeffekt som mediator

To måter det kan trekke opp eller ned. I transmisjon av prisendringer i råvarer og valuta til KPI-vekst, er gjennom formueeffekten.


### Antakelser og hypoteser

### LSTM-modell (hovedmodellen)

Long short-term memory (LSTM) er en type recurrent neural network (RNN) som er spesielt designet for å håndtere sekvensielle data og fange opp langsiktige avhengigheter. RNN har en "hukommelse" som gjør at de kan bruke informasjon fra tidligere tidspunkter i sekvensen for å gjøre prediksjoner. Et problem dette kan skape er long term dependency problem, hvor modellen "husker" data så langt tilbake i tid at den da sliter med å tilegene seg ny informasjon. LSTM løser dette ved å bruke en spesiell arkitektur som inkluderer "gates" for å kontrollere informasjonsflyten og opprettholde relevant informasjon over lengre tidsperioder.

LSTM-modeller bruker vanligvis i dyp læring, og de kan fange opp komplekse, ikke-lineære sammenhenger i dataene. Dette gjør dem godt egnet for å predikere økonomiske tidsserier som KPI-vekst, hvor det kan være mange faktorer som påvirker resultatet og hvor forholdene kan endre seg over tid.
